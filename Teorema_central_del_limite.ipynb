```python
import numpy as np
import numpy.random as npr
import pandas as pd
import seaborn as sns

# Semilla
npr.seed(1)

# DataFrame vacío
df = pd.DataFrame()

# Generar 100 muestras
for i in range(1,101):
    muestras_binomiales = npr.binomial(1000, 0.6, 50)
    col = f"muestras_{i}"
    df[col] = muestras_binomiales

# Calcular la media de cada muestra
df_media = pd.DataFrame(
    df.mean(),
    columns=["muestra de las medias"]
)

# Mostrar medias
df_media

# Histograma
sns.histplot(
    df_media["muestra de las medias"],
    kde=True
)
```
